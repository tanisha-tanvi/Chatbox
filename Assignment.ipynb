```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

url = "https://raw.githubusercontent.com/treselle-systems/customer_churn_analysis/master/WA_Fn-UseC_-Telco-Customer-Churn.csv"
df = pd.read_csv(url)

```

```python
# Convert total charges to numeric (forces empty strings to NaN)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

df.info()
print(df.describe())
```

```python
# Impute missing values
df['TotalCharges_median'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
df['MonthlyCharges_mean'] = df['MonthlyCharges'].fillna(df['MonthlyCharges'].mean())

print(df.head())
```

```python
# Label encoding for binary (Partner: Yes/No)
le = LabelEncoder()
df['Partner_encoded'] = le.fit_transform(df['Partner'])

print(df.head())
```

```python
# One-hot encoding for multi-class (InternetService: DSL, Fiber optic, No)
ohe = OneHotEncoder(sparse_output=False, drop='first')
net_encoded = ohe.fit_transform(df[['InternetService']])
net_df = pd.DataFrame(net_encoded, columns=ohe.get_feature_names_out(['InternetService']))

df = pd.concat([df, net_df], axis=1)

print(df.head())
```
